```python
import numpy as np
```

```python
np.random.seed(42)
```

```python
raw_sales = np.random.randint(10000, 50000, size=(12, 5)).astype(float)
```

```python
raw_sales[2, 1] = np.nan
raw_sales[7, 4] = np.nan
```

```python
traffic = np.random.randint(500, 3000, size=(12, 5))
```

```python
print(" RAW SALES DATA (with missing values):")
print(raw_sales)
```

```python
nan_mask = np.isnan(raw_sales)
```

```python
valid_sales_mean = np.nanmean(raw_sales)
```

```python
cleaned_sales = np.copy(raw_sales)
cleaned_sales[nan_mask] = valid_sales_mean
```

```python
sales_per_customer = cleaned_sales / traffic
```

```python
taxed_sales = np.copy(cleaned_sales)
taxed_sales[5:8, :] *= 0.95
```

```python
branch_annual_totals = np.sum(cleaned_sales, axis=0)
```

```python
monthly_global_totals = np.sum(cleaned_sales, axis=1)
high_performing_months = np.where(monthly_global_totals > 150000)[0]
```

```python
q4_sales = cleaned_sales[9:12, :]
```

```python
print("\n" + "="*50 + "\n")
print(f" Imputed {np.sum(nan_mask)} missing values with global mean: ${valid_sales_mean:,.2f}")
print(f" Total Annual Sales by Branch (A to E): {np.array2string(branch_annual_totals, formatter={'float_kind':lambda x: f'${x:,.2f}'})}")
print(f" Months exceeding $150k global sales (0-indexed): {high_performing_months}")
print(f" Average Sales per Customer overall: ${np.mean(sales_per_customer):,.2f}")
print(f" Q4 Total Revenue: ${np.sum(q4_sales):,.2f}")
```
